DATA CLEANING

Cleaning Weight by converting it from (lbs) -> (Kg)

```python
fifa.Weight = fifa['Weight'].str.replace('lbs','')
```

```python
fifa.Weight = pd.to_numeric(fifa['Weight']) * 0.453592
```

```python
fifa.rename(columns={'Weight':'Weight(Kg)','Height':'Height(cm)'},inplace=True)
```

Convert datatype of jersey from float to int

```python
fifa['Jersey Number'] = fifa['Jersey Number'].fillna(0).astype(int)
```

Convert wages from figures to int

```python
fifa.Wage = fifa.Wage.str.replace('€',"")
fifa.Wage = fifa.Wage.str.replace('K',"000")
fifa.rename(columns={'Wage':'Wages(€)'}, inplace=True)
```

```python
fifa['Wages(€)'] = pd.to_numeric(fifa['Wages(€)'])
```
